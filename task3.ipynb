## if else


## string

```python
str1="Hello"
str2=""

if str1 and str2:
    print("Both strings are non-empty.")
else:
    print("At least one string is empty.")
```

```python
str1 = "hello"
str2 = "world"

if type(str1) == str and type(str2) == str:
    print("both are string")
else:
    print("both are not string")
```

## intiger

```python
x = 5
y = 0

if x != 0 and y != 0:
    print("Both numbers are non-zero.")
else:
    print("At least one number is zero.")


if 9+1 == 10 or 9-1 == 8:
        print("true")
else:
        print("false")
```

## list

```python
list = [10, 20, 30]


if list and 20 in list:
    print("List has elements and contains 20.")
else:
    print("List is empty or 20 not found.")


if not list or len(list) < 5:
    print("List is empty OR has fewer than 5 items.")
```

## Dictionary

```python
person = {"name": "Raman", "age": 25}


if "name" in person and person["name"].istitle():
    print("Name key exists and is properly capitalized.")
else:
    print("Name key missing or not capitalized.")


if person.get("age", 0) > 18 or person.get("role") == "admin":
    print("User is an adult or an admin.")
```
