# Python Predefined Methods for String, Number, and List

### --- STRING METHODS ---

str1 = "Hello, World!"

1. Converts to lowercase
print(str1.lower())          

2. Converts to uppercase
print(str1.upper())                   

3. Capitalizes first letter
print(str1.capitalize())             

4. Capitalizes first letter of each word
print(str1.title())                  

5. Removes leading/trailing whitespace
print(str1.strip())                   

6. Replaces substring
print(str1.replace("World", "Python"))      

7. Finds index of substring (-1 if not found)
print(str1.find("World"))             

8. Splits string into list
print(str1.split(","))               

9. Checks if all characters are alphabetic
print(str1.isalpha())                 

10. Checks if all characters are digits
print(str1.isdigit())                


### --- NUMBER METHODS ---

x = 5
y = -3
f = 3.14159

print(abs(y))                               # Absolute value

print(pow(x, 2))                            # Exponentiation (x^2)

print(round(f, 2))                          # Rounding to 2 decimal places

print(int("15"))                            # Converts string to integer

print(float("12.34"))                       # Converts string to float

### --- LIST METHODS ---

L1 = [3, 1, 4, 1, 5, 9]

L1.append(2)                                # Adds element to end

L1.insert(1, 7)                             # Inserts 7 at index 1

L1.remove(1)                                # Removes first occurrence of value 1

L1.pop()                                    # Removes and returns last item

L1.sort()                                   # Sorts list ascending

L1.reverse()                                # Reverses the list

print(L1.count(1))                          # Counts occurrences of 1

print(L1.index(5))                          # Index of first occurrence of 5
 
L1.clear()                                  # Clears all elements



## STRING METHODS

```python
str1 = "Hello,World"

print(f"lower case of str1 is {str1.lower()}"),

print(f"upper case of str1 is {str1.upper()}"),

print(f"capitalize of str1 is {str1.capitalize()}"),

print(f"tital of str1 is {str1.title()}"),

print(f"strip of str1 is {str1.strip()}"),

print(f"replacement in str1 is {str1.replace('World','python')}"),

print(f"index of world in str1 is {str1.find('World')}"),

print(f"splittes the str1 in {str1.split(',')}"),

print(f"check if the str1 is alpha {str1.isalpha()}"),

print(f"check if the str1 is digit {str1.isdigit()}"),

print(f"check if the str1 is numeric {str1.isnumeric()}")
```
