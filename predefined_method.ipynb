
###  String Methods (`str`)

| Method              | Description                                      |
|---------------------|--------------------------------------------------|
| `lower()`           | Converts string to lowercase                     |
| `upper()`           | Converts string to uppercase                     |
| `strip()`           | Removes whitespace from both ends                |
| `replace(a, b)`     | Replaces substring `a` with `b`                  |
| `find(sub)`         | Returns the index of first occurrence of `sub`  |
| `split(sep)`        | Splits string into list using separator `sep`   |
| `join(iterable)`    | Joins elements of iterable with string as sep   |
| `startswith(sub)`   | Checks if string starts with `sub`              |
| `endswith(sub)`     | Checks if string ends with `sub`                |
| `isdigit()`         | Returns `True` if all characters are digits     |

###  Number Methods (`int`, `float`)

Python’s numbers are primitive types, so they have fewer methods. But here are common built-in functions used with numbers:

| Function             | Description                                        |
|----------------------|----------------------------------------------------|
| `abs(x)`             | Absolute value of `x`                              |
| `pow(x, y)`          | `x` raised to the power `y`                        |
| `round(x, n)`        | Rounds `x` to `n` decimal places                   |
| `int(x)`             | Converts `x` to an integer                         |
| `float(x)`           | Converts `x` to a floating-point number            |
| `divmod(a, b)`       | Returns `(a // b, a % b)` as a tuple               |
| `max(args)`          | Largest item among arguments                       |
| `min(args)`          | Smallest item among arguments                      |

---

###  List Methods (`list`)

| Method              | Description                                        |
|---------------------|----------------------------------------------------|
| `append(x)`         | Adds `x` to the end                                |
| `extend(iterable)`  | Extends list by appending all elements             |
| `insert(i, x)`      | Inserts `x` at position `i`                        |
| `remove(x)`         | Removes first item with value `x`                 |
| `pop([i])`          | Removes and returns item at index `i` (or end)     |
| `index(x)`          | Returns first index of `x`                         |
| `count(x)`          | Counts occurrences of `x`                          |
| `reverse()`         | Reverses the list in place                         |
| `sort()`            | Sorts the list in ascending order                  |
| `clear()`           | Removes all items from the list                    |

## STRING METHODS

```python
str1 = "Hello,World"

print(f"lower case of str1 is {str1.lower()}"),

print(f"upper case of str1 is {str1.upper()}"),

print(f"capitalize of str1 is {str1.capitalize()}"),

print(f"tital of str1 is {str1.title()}"),

print(f"strip of str1 is {str1.strip()}"),

print(f"replacement in str1 is {str1.replace('World','python')}"),

print(f"index of world in str1 is {str1.find('World')}"),

print(f"splittes the str1 in {str1.split(',')}"),

print(f"check if the str1 is alpha {str1.isalpha()}"),

print(f"check if the str1 is digit {str1.isdigit()}"),

print(f"check if the str1 is numeric {str1.isnumeric()}")
```
